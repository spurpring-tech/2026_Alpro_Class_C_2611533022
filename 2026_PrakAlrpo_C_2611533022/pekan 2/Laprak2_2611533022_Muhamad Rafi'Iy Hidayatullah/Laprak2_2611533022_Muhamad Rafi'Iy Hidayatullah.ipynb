### 2.2.1 Program yang berfokus pada penggunaan string

```python
print("Halo, nama anda siapa?")
nama = input()
print("Halo, " + nama + "!")
```

### 2.2.2 Program yang berfokus pada Interger

```python
while True: print(eval(input("Masukkan ekspresi matematika: ")))
```

### 2.2.3 Program yang berfokus pada float

```python
angka = float
angka = input("masukkan angka desimal:")
if angka == float:
    print("ini adalah angka desimal")
else:
    print("ini bukan angka desimal")
    angka = input("masukkan angka desimal:")
```

### 2.2.4 Program yang berfokus pada Boolean

```python
Angka = input("masukkan angka x yang sesuai dengan persamaan 2x*4=10")
if Angka == "5":
    print("Benar! x adalah 5")
else:
    print("Salah! x adalah 5")
```

**Penjelasan :**
1. Bagian pertama kode ini berisi percakapan sederhana
